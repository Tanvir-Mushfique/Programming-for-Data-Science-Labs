Final Laboratory Challenge – Titanic Dataset Analysis

1. How many passengers are in the dataset?

```python
print("Total passengers:", len(df))
```

The Titanic dataset contains a total of 891 passengers.

2. What is the average passenger age?

```python
print("Average passenger age:", df["Age"].mean())
```

The average age of passengers was approximately 29.7 years.

3. Which passenger class has the highest number of passengers?

```python
print(df["Pclass"].value_counts())
```

Third class (Pclass 3) has the highest number of passengers, with 491 passengers.

4. What is the average fare paid by male and female passengers?

```python
print(df.groupby("Sex")["Fare"].mean())
```

Female passengers paid an average fare of approximately 44.48, while male passengers paid approximately 25.52.

5. Which gender has a higher survival rate?


```python
print(df.groupby("Sex")["Survived"].mean())
```

Female passengers had a higher survival rate of approximately 74.20%, compared to male passengers at 18.89%.

6. How does the survival rate differ across passenger classes?

```python
print(df.groupby("Pclass")["Survived"].mean() * 100)
```

First-class passengers had the highest survival rate (62.96%), followed by second-class passengers (47.28%) and third-class passengers (24.24%).
