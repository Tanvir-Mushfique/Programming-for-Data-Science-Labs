```python
marks = [72, 56, 88, 50, 52]

total = sum(marks)
average = total / len(marks)
highest = max(marks)
lowest = min(marks)

passed = 0
for m in marks:
    if m >= 50:
        passed = passed + 1

print("Marks:", marks)
print("Total:", total)
print("Average:", average)
print("Highest mark:", highest)
print("Lowest mark:", lowest)
print("Number of passed courses:", passed)
```

```python
temperatures = (30, 32, 29, 31, 33, 28)

print("Weekend Temperatures:")

for temp in temperatures:
    print(temp)

highest = temperatures[0]
lowest = temperatures[0]

for temp in temperatures:
    if temp > highest:
        highest = temp
    if temp < lowest:
        lowest = temp

print("Highest Temperature:", highest)
print("Lowest Temperature:", lowest)
```

```python
text = input("Enter a word or short sentence: ")

text = text.lower()

characters = 0
vowels = 0
spaces = 0

for ch in text:
    characters = characters + 1

    if ch == " ":
        spaces = spaces + 1

    if ch == "a" or ch == "e" or ch == "i" or ch == "o" or ch == "u":
        vowels = vowels + 1

print("Number of characters:", characters)
print("Number of vowels:", vowels)
print("Number of spaces:", spaces)
```

```python
def calculate_grade(marks):
    average = sum(marks) / len(marks)

    if average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    else:
        return "F"

marks = [85, 75, 80, 70, 90]

grade = calculate_grade(marks)

print("Marks:", marks)
print("Average:", sum(marks) / len(marks))
print("Grade:", grade)
```

```python
products = [
    {"name": "Keyboard", "price": 1200, "quantity": 1},
    {"name": "Mouse", "price": 500, "quantity": 2},
    {"name": "USB Cable", "price": 200, "quantity": 1},
    {"name": "Speakers", "price": 900, "quantity": 1}
]

total = 0

for product in products:
    cost = product["price"] * product["quantity"]
    total = total + cost

def apply_discount(total):
    if total >= 2000:
        return total * 0.90
    else:
        return total

final_total = apply_discount(total)

print("Total Cost:", total)
print("Final Cost:", final_total)
```

```python
students = [
    {"name": "Rahim", "department": "CSE", "marks": 85, "attendance": 92},
    {"name": "Karim", "department": "CSE", "marks": 48, "attendance": 76},
    {"name": "Sara", "department": "DS", "marks": 91, "attendance": 95},
    {"name": "Nadia", "department": "DS", "marks": 67, "attendance": 84},
    {"name": "Hasan", "department": "CSE", "marks": 73, "attendance": 78}
]

total = 0
passed = 0
highest = students[0]

for student in students:
    print(student["name"], "-", student["marks"])

    total = total + student["marks"]

    if student["marks"] >= 50:
        passed = passed + 1

    if student["marks"] > highest["marks"]:
        highest = student

average = total / len(students)

print("Passed Students:", passed)
print("Class Average:", average)
print("Highest Mark Student:", highest["name"])
print("Highest Mark:", highest["marks"])
```

```python
products = [
    {"name": "Notebook", "price": 120, "quantity": 3},
    {"name": "Pen", "price": 20, "quantity": 5},
    {"name": "USB Drive", "price": 850, "quantity": 2},
    {"name": "Mouse", "price": 650, "quantity": 1}
]

total_value = 0
highest_product = products[0]
highest_total = 0

for product in products:
    line_total = product["price"] * product["quantity"]

    print(product["name"], "Line Total:", line_total)

    total_value = total_value + line_total

    if line_total > highest_total:
        highest_total = line_total
        highest_product = product

print("Total Inventory Value:", total_value)
print("Highest Line Total Product:", highest_product["name"])
print("Highest Line Total:", highest_total)
```
